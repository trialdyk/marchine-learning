# Lab 1 - Ekstraksi Fitur pada Data Tabular

## Langkah 1 - Import Library

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import chi2, SelectKBest, RFE
from sklearn.metrics import accuracy_score, classification_report

print('Library loaded OK')
```

## Langkah 2 - Load Data dan Pengelompokan Data

```python
# Load Data
df = pd.read_csv("Titanic-Dataset.csv")

# Pisahkan Survived
y = df["Survived"].astype(int)
X = df.drop(columns=["Survived"])

# Buat list variabel numerik dan kategorikal
# Name tidak akan digunakan karena tidak relevan
num_cols = ["Age", "SibSp", "Parch", "Fare"]
cat_cols = ["Pclass", "Sex", "Embarked"]

print('X shape:', X.shape)
print('y shape:', y.shape)
```

## Langkah 3 - Ekstraksi Fitur (Pipeline)

Pipeline = model proses yang bisa dipakai berulang kali. Numerik: imputasi median + scaling. Kategorikal: imputasi modus + one-hot.

```python
# Ekstraksi Fitur dengan Pipeline

# Data Numerik
num_tf = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Data Kategorikal
cat_tf = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

print('Pipeline preprocessing siap')
```

## Langkah 3.1 - Feature Construction

Membuat fitur baru `FamilySize = SibSp + Parch + 1`.

```python
# Buat Fitur FamilySize
X["FamilySize"] = X["SibSp"].fillna(0) + X["Parch"].fillna(0) + 1

# Tambahkan FamilySize pada kelompok numerikal
preprocess = ColumnTransformer([
    ("num", num_tf, num_cols + ["FamilySize"]),
    ("cat", cat_tf, cat_cols),
])

X[["SibSp", "Parch", "FamilySize"]].head()
```

## Langkah 4 - Seleksi Fitur (SelectKBest / ANOVA)

Univariate statistics: pilih k fitur terbaik berdasarkan skor ANOVA F-value.

```python
from sklearn.feature_selection import f_classif

selector_filter = SelectKBest(score_func=f_classif, k=5)

# Buat pipeline final (BELUM MEMPROSES DATA, baru rangkaian)
pipe_filter = Pipeline([
    ("prep", preprocess),   # preprocessing
    ("sel", selector_filter),  # seleksi fitur
    ("clf", LogisticRegression(max_iter=1000))  # model sederhana: Logistic Regression
])

print('Pipeline final siap')
```

## Langkah 5 - Uji dengan Model

```python
# Pelatihan dan uji model
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)

pipe_filter.fit(X_train, y_train)
pred = pipe_filter.predict(X_test)

print("=== Filter (ANOVA) + LR ===")
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

### Inspeksi: fitur apa yang dipakai?

```python
# 1) Nama fitur setelah preprocess
feat_names = pipe_filter.named_steps["prep"].get_feature_names_out()
print("Nama fitur:", feat_names)
print("\n")

# 2) Mask & skor fitur terpilih (SelectKBest)
sel = pipe_filter.named_steps["sel"]
mask = sel.get_support()
selected_names = feat_names[mask]
selected_scores = sel.scores_[mask]

top = sorted(zip(selected_names, selected_scores), key=lambda t: t[1], reverse=True)[:10]
print("Top fitur:", top)
```

**Kesimpulan modul:** akurasi ~0.78 didapat dari fitur `Sex_female`, `Sex_male`, `Pclass_3`, `Pclass_1`, `Fare`.

Coba eksperimen: ubah `k` pada SelectKBest atau ganti metode seleksi (mis. RFE, chi2) untuk hasil terbaik.
