# Lab 2 - Data Imputation

## Langkah 0 - Persiapan

```python
import pandas as pd
import numpy as np

print('Library loaded OK')
```

## Langkah 1 - Memuat Data

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

## Langkah 2 - Inspeksi Data

```python
# Inspeksi singkat: cek data hilang sebelum imputasi
df.info()
print()
df.isnull().sum()
```

## Langkah 3 - Data Imputation

```python
# Age  -> strategi mean (data nominal/kuantitatif)
df['Age'].fillna(value=df['Age'].mean(), inplace=True)

# Cabin -> substitusi 'DECK' (asumsi: tanpa nomor kabin = penumpang di dek)
df['Cabin'].fillna(value='DECK', inplace=True)

# Embarked -> modus (mode), karena data nominal dan cuma 2 nilai hilang
# Catatan: modul asli menulis mode (tanpa [0]) yang akan error.
# Mode sebenarnya adalah Seri/DataFrame, nilainya diambil dengan index [0].
df['Embarked'].fillna(value=df['Embarked'].mode()[0], inplace=True)

print('Imputasi selesai')
```

## Langkah 4 - Validasi Hasil

```python
# Cek kembali apakah data hilang sudah terisi
df.info()
print()
df.isnull().sum()
```

```python
# Validasi langsung: cek nilai yang baru diisi (Age, Cabin, Embarked)
print('Age mean  :', df['Age'].mean())
print('Cabin mode:', df['Cabin'].value_counts().idxmax())
print('Emb modus :', df['Embarked'].value_counts().idxmax())
print()
df[['Age', 'Cabin', 'Embarked']].head()
```

## (Opsional) Simpan Hasil

```python
# Simpan hasil imputasi untuk tahapan selanjutnya
df.to_csv('Titanic-Dataset-Fixed.csv', index=False)
print('Tersimpan: Titanic-Dataset-Fixed.csv')
```
