# Lab 3 - Seleksi Fitur

Memilih fitur dari data Titanic yang sudah diimputasi (Lab 2), lalu melakukan encoding dan standarisasi.

## Langkah 0 - Load Library

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

print('Library loaded OK')
```

## Langkah 1 - Load Data

```python
from google.colab import drive
drive.mount('/content/drive')
path = '/content/drive/MyDrive/MachineLearning/Titanic-Dataset-Fixed.csv'

df = pd.read_csv(path)
df.head()
```

## Langkah 2 - Slice Data (pilih fitur)

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

## Langkah 3 - Encoding

```python
le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])
df['Cabin'] = le.fit_transform(df['Cabin'])

print('Encoding selesai (Sex & Cabin -> label)')
```

## Langkah 4 - Verifikasi Hasil Encoding

```python
df.head()
```

## Langkah 5 - Standarisasi

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])

print('Standarisasi Age selesai')
```
